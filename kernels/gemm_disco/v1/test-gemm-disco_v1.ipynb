```python
# Imports
import pandas as pd
from random import randint
from src import *
from src.simulator import SIMULATOR
```

```python
sim = SIMULATOR()

DEBUG = 1
# --------------------------------------------
#                DATA 
# --------------------------------------------
# DISCO-CGRA Configuration
nRCs = 4
nElementsPerVWRSlice = 32
nColsCGRA = 2

#Blocks A:6x128, B:128x32, C:6x32
BLOCK_ROWS = 6

# Our test
ROWS_A = 12 # Multiple of 6
COLS_A = 128
COLS_B = 32

ALPHA = 1
BETA = 0

matrix_A = np.array([(i % 10) * pow(-1, i) for i in range(ROWS_A*COLS_A)])
#matrix_A = np.eye(ROWS_A, COLS_A, dtype=int).flatten()
matrix_B = np.array([(i % 20) * pow(-1, i) for i in range(COLS_A*COLS_B)])
matrix_C = np.array([(i % 30) * pow(-1, i) for i in range(ROWS_A*COLS_B)])
matrix_B_t = (matrix_B.reshape(COLS_A, COLS_B)).T.flatten()

# --------------------------------------------
#               KERNEL CONFIGURATION
# --------------------------------------------
kernel_path = './kernels/gemm_disco_wo_mac/'
kernel_number = 1 
column_usage = [True, True] 
nInstrPerCol = 19
imem_add_start = 0 
srf_spm_addres = 0 
version=""

sim.kernel_config(column_usage, nInstrPerCol, imem_add_start, srf_spm_addres, kernel_number)
```

```python
def printAsMatrix(array, rows, cols):
    for i in range(rows):
        print(array[i * cols:(i + 1) * cols])
```

```python
# Default SPM lines
srf_spm_line = 0
b_spm_line_ini = 1
b_spm_line_end = b_spm_line_ini + 32
a_spm_line_ini_buff1 = b_spm_line_end + 1
a_spm_line_end_buff1 = a_spm_line_ini_buff1 + BLOCK_ROWS
c_spm_line_ini_buff1 = a_spm_line_end_buff1 + 1
c_spm_line_end_buff1 = c_spm_line_ini_buff1 + BLOCK_ROWS
a_spm_line_ini_buff2 = c_spm_line_end_buff1 + 1
a_spm_line_end_buff2 = a_spm_line_ini_buff2 + BLOCK_ROWS
c_spm_line_ini_buff2 = a_spm_line_end_buff2 + 1
c_spm_line_end_buff2 = c_spm_line_ini_buff2 + BLOCK_ROWS

def fillSPMwithB_t(matrixB_t, COLS_A, COLS_B):
    b_spm_line = b_spm_line_ini
    for r in range(COLS_B):
        sim.setSPMLine(b_spm_line, matrixB_t[r*COLS_A: (r+1)*COLS_A].copy())
        b_spm_line+=1

def fillBuffWithAandCandSRF(srf, matrix_A, matrix_C, COLS_A, COLS_B, nBuf, firstRow):
    a_spm_line = a_spm_line_ini_buff1 if nBuf==1 else a_spm_line_ini_buff2
    c_spm_line = c_spm_line_ini_buff1 if nBuf==1 else c_spm_line_ini_buff2

    for i in range(BLOCK_ROWS):
        sim.setSPMLine(a_spm_line, matrix_A[(firstRow + i)*COLS_A: (firstRow + i +1)*COLS_A].copy())
        a_spm_line+=1
        sim.setSPMLine(c_spm_line, np.concatenate([matrix_C[(firstRow + i)*COLS_B: (firstRow + i +1)*COLS_B], np.zeros(128 - COLS_B,dtype=matrix_C.dtype)]).copy())
        c_spm_line+=1
    
    sim.setSPMLine(srf_spm_line, srf.copy())

def getCFromBuff(matrix_C, COLS_B, nBuf, firstRow):
    c_spm_line = c_spm_line_ini_buff1 if nBuf==1 else c_spm_line_ini_buff2
    outs = []

    for i in range(BLOCK_ROWS):
        out = sim.getSPMLine(c_spm_line)
        outs.append(out[:COLS_B].copy())
        c_spm_line+=1
    return outs

def fillSRF(nBuf):
    # --------------------------------------------
    # SRF0 = SPMA
    # SRF1 = SPMC
    # SRF2 = rC
    # SRF3 = alpha
    # SRF4 = beta
    # --------------------------------------------

    # Default SRF values
    srf = [0 for i in range(N_ELEMS_PER_VWR)]

    # Col 0
    srf[0]  = a_spm_line_ini_buff1 if nBuf==1 else a_spm_line_ini_buff2
    srf[1]  = c_spm_line_ini_buff1 if nBuf==1 else c_spm_line_ini_buff2
    srf[2]  = int(BLOCK_ROWS/2) 
    srf[3]  = ALPHA 
    srf[4]  = BETA
    srf[5]  = 0      # Not used
    srf[6]  = 0      # Not used
    srf[7]  = 0      # Not used
    # Col 1
    srf[8]  = a_spm_line_ini_buff1 + int(BLOCK_ROWS/2) if nBuf==1 else a_spm_line_ini_buff2 + int(BLOCK_ROWS/2)
    srf[9]  = c_spm_line_ini_buff1 + int(BLOCK_ROWS/2) if nBuf==1 else c_spm_line_ini_buff2 + int(BLOCK_ROWS/2)
    srf[10] = int(BLOCK_ROWS/2) 
    srf[11] = ALPHA
    srf[12] = BETA
    srf[13] = 0      # Not used
    srf[14] = 0      # Not used
    srf[15] = 0      # Not used
    return srf
    
```

```python
# --------------------------------------------
#              COMPILE ASM TO HEX
# --------------------------------------------
sim.compileAsmToHex(kernel_path, kernel_number, version=version)

# --------------------------------------------
#                 LOAD KERNEL
# --------------------------------------------

# This needs the hex instructions, if you don't provide them, generate then compiling the asm
sim.kernel_load(kernel_path, version=version + "_autogen", kernel_number=kernel_number)
```

Finally, we load the kernel into the internal memory of the specialized units and run it.

```python
def runKernel(matrix_A, matrix_B_t, matrix_C, ROWS_A, COLS_A, COLS_B, max_iter=3000):
    # --------------------------------------------
    #               SIMULATE EXECUTION
    # --------------------------------------------
    show_lcu = []
    show_srf = []
    show_lsu = []
    show_rcs = [[],[],[],[]]
    show_mxcu = []
    display_ops = [show_lcu, show_lsu, show_mxcu, show_rcs, show_srf]

    # LOAD STATIC VALUES INTO SPM
    fillSPMwithB_t(matrix_B_t, COLS_A, COLS_B)
    sim.displaySPMLine(1) # B
    sim.displaySPMLine(2) # B
    # BUFFERING AND COMPUTATION
    outs = []
    for i in range(ROWS_A // BLOCK_ROWS):
        print(f"--- Processing block {i} ---")
        sim.kernel_load(kernel_path, version=version + "_autogen", kernel_number=kernel_number)
        sim.displaySPMLine(1) # B
        sim.displaySPMLine(2) # B
        firstRow = i * BLOCK_ROWS
        nBuf = 1 if i %2 == 0 else 2

        # Fill buffer with A and C and SRF
        srf = fillSRF(nBuf)
        fillBuffWithAandCandSRF(srf, matrix_A, matrix_C, COLS_A, COLS_B, nBuf, firstRow)
        if DEBUG:
            sim.displaySPMLine(0) # SRF
            sim.displaySPMLine(1) # B0
            sim.displaySPMLine(2) # B1
            sim.displaySPMLine(a_spm_line_ini_buff1) # A0
            sim.displaySPMLine(c_spm_line_ini_buff1) # C0

        # Launch kernel
        sim.run(kernel_number, display_ops=display_ops, max_iter=max_iter)

        # Get C from buffer
        outs.append(getCFromBuff(matrix_C, COLS_B, nBuf, firstRow))
    return outs
```

We can check it more rigorously. We can define our function in python and check that the output matches the CGRA output.

```python
def gemm (in_A, in_B, nRowsA, nColsA, nColsB, alpha, beta):
    out = np.zeros(nRowsA*nColsB)
    for i in range(nRowsA):
        for j in range(nColsB):
            sum = 0
            for k in range(nColsA):
                sum += int(alpha * in_A[i*nColsA + k] * in_B[k*nColsB + j])
            out[i*nColsB + j] = sum + beta*out[i*nColsB + j]
    return [int(elem) for elem in out]
```

```python
cpu_out = gemm(matrix_A, matrix_B, ROWS_A, COLS_A, COLS_B, ALPHA, BETA)
printAsMatrix(cpu_out, ROWS_A, COLS_B)
```

```python
# Run kernel
disco_out_aux = runKernel(matrix_A, matrix_B_t, matrix_C, ROWS_A, COLS_A, COLS_B, max_iter=30000)
disco_out = [int(x) for row in disco_out_aux for arr in row for x in arr]
```

```python
# Verify results
errors = 0
for i in range(len(cpu_out)):
    if cpu_out[i] != disco_out[i]:
        errors+=1
    
if errors == 0:
    print("The result is correct!")
else:
    print("Oops, something went wrong. There are " + str(errors) + " errors (out of " + str(len(cpu_out)) + " elements).")
    print("DISCO out:")
    printAsMatrix(disco_out, ROWS_A, COLS_B)
    print("Expected result:")
    printAsMatrix(cpu_out, ROWS_A, COLS_B)
```
